## Camada Bronze
Carrega o PARTICIPANTES_2025.csv do volume, sem alterações, e acrescenta a data da ingestão e o arquivo de origem.

```sql
CREATE OR REPLACE TABLE workspace.bronze.participantes AS
SELECT *,
       current_timestamp()      AS data_ingestao,
       'PARTICIPANTES_2025.csv' AS arquivo_origem
FROM read_files(
  '/Volumes/workspace/bronze/arquivos/PARTICIPANTES_2025.csv',
  format => 'csv', header => true, sep => ';', encoding => 'ISO-8859-1'
);
```

```sql
SELECT COUNT(*) FROM workspace.bronze.participantes;
```

```sql
SELECT * FROM workspace.bronze.participantes LIMIT 10;
```

```sql
DESCRIBE workspace.bronze.participantes;
```

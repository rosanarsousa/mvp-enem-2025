```sql
CREATE OR REPLACE TABLE workspace.gold.fato_inscricao AS
SELECT nu_inscricao,
       sg_uf,
       municipio_prova,
       cod_faixa_etaria,
       cod_situacao_conclusao,
       cod_tipo_ensino
FROM workspace.silver.participantes;
```

```sql
SELECT COUNT(*) AS total_fato FROM workspace.gold.fato_inscricao;
```

```sql
SELECT * FROM workspace.gold.fato_inscricao LIMIT 10;
```

```sql
SELECT
  (SELECT COUNT(*) FROM workspace.gold.fato_inscricao f
     LEFT JOIN workspace.gold.dim_uf d ON f.sg_uf = d.sg_uf
     WHERE d.sg_uf IS NULL) AS sem_par_uf,
  (SELECT COUNT(*) FROM workspace.gold.fato_inscricao f
     LEFT JOIN workspace.gold.dim_faixa_etaria d ON f.cod_faixa_etaria = d.cod_faixa_etaria
     WHERE d.cod_faixa_etaria IS NULL) AS sem_par_faixa_etaria,
  (SELECT COUNT(*) FROM workspace.gold.fato_inscricao f
     LEFT JOIN workspace.gold.dim_situacao_conclusao d ON f.cod_situacao_conclusao = d.cod_situacao_conclusao
     WHERE d.cod_situacao_conclusao IS NULL) AS sem_par_situacao,
  (SELECT COUNT(*) FROM workspace.gold.fato_inscricao f
     LEFT JOIN workspace.gold.dim_tipo_ensino d ON f.cod_tipo_ensino = d.cod_tipo_ensino
     WHERE d.cod_tipo_ensino IS NULL) AS sem_par_tipo_ensino;
```

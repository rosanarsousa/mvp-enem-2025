## Qualidade de dados
Verifica completude, unicidade, consistência e acurácia das 6 colunas do estudo, na camada bronze, antes de qualquer tratamento.

```sql
SELECT COUNT(*) AS total,
       COUNT(*) - COUNT(NU_INSCRICAO)       AS vazios_inscricao,
       COUNT(*) - COUNT(SG_UF_PROVA)        AS vazios_uf,
       COUNT(*) - COUNT(NO_MUNICIPIO_PROVA) AS vazios_municipio,
       COUNT(*) - COUNT(TP_FAIXA_ETARIA)    AS vazios_faixa_etaria,
       COUNT(*) - COUNT(TP_ST_CONCLUSAO)    AS vazios_situacao,
       COUNT(*) - COUNT(TP_ENSINO)          AS vazios_tipo_ensino
FROM workspace.bronze.participantes;
```

```sql
SELECT COUNT(*) AS linhas,
       COUNT(DISTINCT NU_INSCRICAO) AS inscricoes_unicas
FROM workspace.bronze.participantes;
```

```sql
SELECT TP_FAIXA_ETARIA, COUNT(*) AS qtd
FROM workspace.bronze.participantes
GROUP BY TP_FAIXA_ETARIA ORDER BY TP_FAIXA_ETARIA;
```

```sql
SELECT TP_ST_CONCLUSAO, COUNT(*) AS qtd
FROM workspace.bronze.participantes
GROUP BY TP_ST_CONCLUSAO ORDER BY TP_ST_CONCLUSAO;
```

```sql
SELECT TP_ENSINO, COUNT(*) AS qtd
FROM workspace.bronze.participantes
GROUP BY TP_ENSINO ORDER BY TP_ENSINO;
```

```sql
SELECT COUNT(DISTINCT SG_UF_PROVA) AS qtd_ufs
FROM workspace.bronze.participantes;
```

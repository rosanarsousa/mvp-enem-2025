## Catálogo de dados
Descrição de cada tabela e coluna no Unity Catalog, com domínio de valores e linhagem (origem de cada campo).

```sql
COMMENT ON TABLE workspace.gold.fato_inscricao IS 'Tabela fato do modelo estrela. Grão: uma linha por inscrito no Enem 2025 (4.810.772 registros). Fonte: PARTICIPANTES_2025.csv (Inep), via bronze.participantes e silver.participantes.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN nu_inscricao COMMENT 'Número de inscrição anonimizado pelo Inep. Chave única, sem duplicatas. Origem: NU_INSCRICAO.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN sg_uf COMMENT 'Sigla da UF de aplicação da prova (27 valores). Chave para dim_uf. Origem: SG_UF_PROVA, com TRIM e UPPER.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN municipio_prova COMMENT 'Município de aplicação da prova. Base para a estratégia de Uber Ads. Origem: NO_MUNICIPIO_PROVA, com TRIM.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN cod_faixa_etaria COMMENT 'Código da faixa etária (0 a 20, ver dim_faixa_etaria). Origem: TP_FAIXA_ETARIA.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN cod_situacao_conclusao COMMENT 'Código da situação de conclusão do ensino médio (0 a 4, ver dim_situacao_conclusao). Origem: TP_ST_CONCLUSAO.';
ALTER TABLE workspace.gold.fato_inscricao ALTER COLUMN cod_tipo_ensino COMMENT 'Código do tipo de ensino (0 a 2, ver dim_tipo_ensino). O código 0 = não informado foi atribuído na silver aos 3.080.608 registros vazios. Origem: TP_ENSINO.';
```

```sql
COMMENT ON TABLE workspace.gold.dim_uf IS 'Dimensão de unidades da federação: 27 UFs com nome, região e presença atual da marca. Criada manualmente com base na divisão regional do IBGE.';
ALTER TABLE workspace.gold.dim_uf ALTER COLUMN sg_uf COMMENT 'Sigla da UF (chave, 27 valores).';
ALTER TABLE workspace.gold.dim_uf ALTER COLUMN nome_uf COMMENT 'Nome da unidade da federação.';
ALTER TABLE workspace.gold.dim_uf ALTER COLUMN regiao COMMENT 'Região do Brasil: Norte, Nordeste, Centro-Oeste, Sudeste ou Sul.';
ALTER TABLE workspace.gold.dim_uf ALTER COLUMN presenca_marca COMMENT 'Sim = UF onde o grupo já atua (BA e MG); Não = demais. Regra de negócio criada na camada gold.';

COMMENT ON TABLE workspace.gold.dim_faixa_etaria IS 'Dimensão de faixa etária: traduz os códigos de TP_FAIXA_ETARIA conforme o dicionário do Inep 2025 (21 linhas).';
ALTER TABLE workspace.gold.dim_faixa_etaria ALTER COLUMN cod_faixa_etaria COMMENT 'Código da faixa etária (0 a 20). 0 = não informado.';
ALTER TABLE workspace.gold.dim_faixa_etaria ALTER COLUMN faixa_etaria COMMENT 'Descrição da faixa etária conforme o dicionário do Inep, de Menor de 17 anos a Maior de 70 anos.';
ALTER TABLE workspace.gold.dim_faixa_etaria ALTER COLUMN publico_alvo COMMENT 'Sim = códigos 1 a 5 (menor de 17 a 20 anos), público prioritário da campanha; Não = demais. Regra de negócio criada na camada gold.';

COMMENT ON TABLE workspace.gold.dim_situacao_conclusao IS 'Dimensão de situação de conclusão do ensino médio: traduz os códigos de TP_ST_CONCLUSAO conforme o dicionário do Inep 2025 (5 linhas).';
ALTER TABLE workspace.gold.dim_situacao_conclusao ALTER COLUMN cod_situacao_conclusao COMMENT 'Código da situação (0 a 4). 0 = não informado.';
ALTER TABLE workspace.gold.dim_situacao_conclusao ALTER COLUMN situacao_conclusao COMMENT 'Descrição da situação de conclusão do ensino médio conforme o dicionário do Inep.';
ALTER TABLE workspace.gold.dim_situacao_conclusao ALTER COLUMN grupo_situacao COMMENT 'Agrupamento para análise: Já concluiu (1), Cursando o ensino médio (2 e 3), Não concluiu e não está cursando (4), Não informado (0). Regra criada na camada gold.';

COMMENT ON TABLE workspace.gold.dim_tipo_ensino IS 'Dimensão de tipo de ensino: traduz os códigos de TP_ENSINO conforme o dicionário do Inep 2025 (3 linhas).';
ALTER TABLE workspace.gold.dim_tipo_ensino ALTER COLUMN cod_tipo_ensino COMMENT 'Código do tipo de ensino (0 a 2). 0 = não informado.';
ALTER TABLE workspace.gold.dim_tipo_ensino ALTER COLUMN tipo_ensino COMMENT 'Descrição: Ensino regular, Educação especial – modalidade substitutiva ou Não informado.';
```
